```python
import pandas as pd

disease_df = pd.read_csv("data/Diseases_and_Symptoms_dataset.csv")
desc_df = pd.read_csv("data/description.csv")
med_df = pd.read_csv("data/medications.csv")
prec_df = pd.read_csv("data/precautions.csv")
workout_df = pd.read_csv("data/workout.csv")
diet_df = pd.read_csv("data/diets.csv")
```

## Preprocessing

```python
# ---Strip character, lower and upper case character---
def clean_disease(series):
    return series.str.strip().str.lower()

disease_df["diseases"] = clean_disease(disease_df["diseases"])
desc_df["Disease"] = clean_disease(desc_df["Disease"])
med_df["Disease"] = clean_disease(med_df["Disease"])
prec_df["Disease"] = clean_disease(prec_df["Disease"])
workout_df["Disease"] = clean_disease(workout_df["Disease"])
diet_df["Disease"] = clean_disease(diet_df["Disease"])
```
